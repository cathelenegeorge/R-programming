```python
%%R
#Linear Regression
height<-c(175,178,182,165,163,154)
weight<-c(65,67,75,70,69,50)
relation <- lm(weight~height)
print(relation)
print(summary(relation))
df<-data.frame(height=160)
y = predict(relation,df)
print(y)
plot(weight,height,col = "blue",main ="Height vs Weight Regression Analysis")
```

```python
%%R
#Multiple Linear Regression
input <- mtcars[,c("mpg","disp","hp","wt")]
model <- lm(mpg~disp+hp+wt,data = input)
print(model)
print(summary(model))
df <- data.frame(disp=200,hp=100,wt=3.5)
y = predict(model,df)
print(y)
plot(input$mpg,input$disp+input$hp+input$wt,col = "red",main ="MPG Analysis")

```

```python
%%R
## Logistic Regression
input<-mtcars[,c("am","cyl","hp","wt")]
model = glm(am~cyl+hp+wt,data = input,family = "binomial")
print(model)
print(summary(model))
df<-data.frame(cyl=4,hp=100,wt=3.5)
y = predict(model,df)
print(y)
plot(input$am,input$cyl+input$hp+input$wt,col = "blue",main ="AM Analysis")
```

```python
%%R
##Normal Distribution
#dnorm
x<-c(68,72,59,86,88,92,64,78,66,79)
y<-dnorm(x,mean(x),sd(x))
print("Using Dnorm Function")
print(y)
plot(x,y)
```

```python
%%R
z <-pnorm(x,mean(x),sd(x))
print("Using Pnorm Function")
print(z)
plot(x,z)
```

```python
%%R
a<-qnorm(z,mean(x),sd(x))
print("Using Qnorm Function")
print(a)
plot(x,a)
```

```python
%%R
#Binomial Distribution
#dbinom
x <-c(3,4,5,6)
y <-dbinom(x,size =10,prob = 0.5)
print("Using Dbinom Function")
print(y)
plot(x,y,col = "red",main = "Binomial Distribution using Dbinom")
```

```python
%%R
x <-c(3,4,5,6)
y <-pbinom(x,size =10,prob = 0.5)
print("Using Pbinom Function")
print(y)
plot(x,y,col = "red",main = "Binomial Distribution using Pbinom")
```

```python
%%R
y <-pbinom(x,size =10,prob = 0.5)
z <- qbinom(y,size =10,prob =0.5)
print("Using Qbinom Function")
print(z)
plot(z,y,col = "red",main = "Binomial Distribution using Qbinom")
```

```python
%%R
x<-rbinom(8,15,prob =0.2)
hist(x,col = "blue",main = "Binomial Distribution using Rbinom")
```

```python
%%R
##Time Series Analysis
rainfall<-c(799,1873,2345,322,250,800,1654,1764,2145,992)
timeseries.object <-ts(rainfall,start = c(2020,1),end = c(2020,10),frequency = 12)
print(timeseries.object)
plot(timeseries.object)
```

```python
%%R
##Multiple Time Series Analysis
rainfall1<-c(799,1873,2345,322,250,800,1654,1764,2145,992)
rainfall2<-c(800,1654,1764,2145,992,799,1873,2345,322,250)
rainfall<-matrix(c(rainfall1,rainfall2),nrow = 12)
timeseries.object <-ts(rainfall,start = c(2020,1),end = c(2020,10),frequency = 12)
print(timeseries.object)
plot(timeseries.object)
```

```python
%%R
##Analysis of Variance
pain <-c(2,3,4,5,7,9,8,2,4,
         4,5,6,7,9,3,5,7,2,
         4,4,5,5,6,7,7,9,9)
drug <-c(rep("A",9),rep("B",9),rep("C",9))
migraine<-data.frame(drug,pain)
print(migraine)
boxplot(pain~drug,data = migraine,main = "Migraine Analysis")
model = aov(pain~drug,data = migraine)
print(model)
print(summary(model))
```
