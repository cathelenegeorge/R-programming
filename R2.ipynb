```python
%%R
## Data Visualisation In R
#Pie Chart
x <- c(21,64,68,27,32)
labels <- c("A","B","C","D","E")
perce<-round(100*x/sum(x),1)
pie(x,labels = perce,col = heat.colors(length(x)),main = "Pie Chart for given data")
legend("topright",legend = labels,col = heat.colors(length(x)),pch = 9)
```

```python
%%R
#barchart
x<-c(30,50,80,100,45,60,25)
y<-c("a","b","c","d","e","f","g")
barplot(x,names.arg = y,xlab = "alphas",ylab = "values",col = "blue",main="Bar chart for given data")
```

```python
%%R
input <- mtcars[,c('mpg','cyl')]
print(head(input))
boxplot(mpg~cyl,xlab= "No.of cylinders",ylab = "Miles Per Gallon",
        data = input,notch = TRUE,varwidth =TRUE,main = "Box plot",col = "red")
```

```python
%%R
##Histograms
x<-c(20,20,70,70,40,80,50,60)
hist(x,main = "Histogram Plot for given data",xlab="Value",ylab = "Frequency",col="orange")
```

```python
%%R
#Performing Operations on CSV files
#Analyzing the csv file
data<-read.csv("student_details.csv")
print(is.data.frame(data))
print(nrow(data))
cat("\n")
print(ncol(data))
cat("\n")
print(head(data))
cat("\n")
print(tail(data))
cat("\n")
print(summary(data))
cat("\n")
print(dim(data))
cat("\n")
##Fetching Info from csv file
#Get the maximum age
age<- max(data$Age)
print(age)
cat("\n")
#Get the details for person with maximum age
retval <- subset(data,Age == max(data$Age))
print(retval)
cat("\n")
phy<-subset(data,Major == "Physics")
print(phy)
cat("\n")
pep<-subset(data,Major == "Physics" & Age >20)
print(pep)
#Writing a csv file
write.csv(pep,"pep.csv")
ip<-read.csv("pep.csv")
print(ip)
#Dealing with missing values
summary(data)
x = any(is.na(data))
print(x)
if (x == TRUE){
    Clean = na.omit(data)#To remove that particular na coloumn
    print(Clean)
    data$Age[is.na(data)]<-mean(data$Age,na.rm = TRUE)#To replace with mean value
}else{
    print("No missing values")
}

```

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("student_details.csv")

# Convert to XML
df.to_xml("student_details.xml", index=False)

# Optional: Download the XML file
from google.colab import files
files.download("student_details.xml")
```

```python
%%R
# Install and load the XML package
if (!requireNamespace("XML", quietly = TRUE)) {
  install.packages("XML")
}
library(XML)

#Performing Operations on XML files
xml<-xmlParse(file = "student_details.xml")
print(xml)
```

```python
%%R
## Extracting root nodes and their sizes
root<-xmlRoot(xml)
print(root)

rootsiz<-xmlSize(root)
print(rootsiz)
print(root[1])
```

```python
%%R
##Converting xml file to dataframe
xmldf<-xmlToDataFrame(xml)
print(xmldf)
```
